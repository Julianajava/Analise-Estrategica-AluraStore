### Importação dos dados



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja.head()
```

#1. Análise do faturamento



```python
faturamento_loja1 = loja['Preço'].sum()
faturamento_loja2 = loja2['Preço'].sum()
faturamento_loja3 = loja3['Preço'].sum()
faturamento_loja4 = loja4['Preço'].sum()

print("Faturamento Loja 1:", faturamento_loja1)
print("Faturamento Loja 2:", faturamento_loja2)
print("Faturamento Loja 3:", faturamento_loja3)
print("Faturamento Loja 4:", faturamento_loja4)
```

```python
import pandas as pd

faturamentos = pd.Series(
    [faturamento_loja1, faturamento_loja2, faturamento_loja3, faturamento_loja4],
    index=['Loja 1', 'Loja 2', 'Loja 3', 'Loja 4']
)

faturamentos
```

# 2. Vendas por Categoria

```python
categoria_loja1 = loja['Categoria do Produto'].value_counts()
categoria_loja2 = loja2['Categoria do Produto'].value_counts()
categoria_loja3 = loja3['Categoria do Produto'].value_counts()
categoria_loja4 = loja4['Categoria do Produto'].value_counts()

print("Loja 1")
print(categoria_loja1)

print("\nLoja 2")
print(categoria_loja2)

print("\nLoja 3")
print(categoria_loja3)

print("\nLoja 4")
print(categoria_loja4)
```

```python
import matplotlib.pyplot as plt

plt.figure()
loja['Categoria do Produto'].value_counts().plot(kind='bar')
plt.title('Produtos Vendidos por Categoria - Loja 1')
plt.xlabel('Categoria')
plt.ylabel('Quantidade Vendida')
plt.show()
```

## 1️⃣ Introdução

Este relatório apresenta uma análise comparativa das quatro lojas da rede **Alura Store**, com o objetivo de identificar qual unidade apresenta menor eficiência operacional e estratégica.

Foram analisados os seguintes indicadores:

- 💰 Faturamento total  
- 🛍 Categorias e produtos mais/menos vendidos  
- ⭐ Média de avaliação dos clientes  
- 🚚 Custo médio de frete  

As análises foram realizadas utilizando Python (Pandas e Matplotlib).

## 2️⃣ Análise dos Indicadores

### 💰 Faturamento

A análise do faturamento revelou diferenças significativas entre as lojas.  
Observou-se que a **Loja X** apresentou o menor volume total de receita.

O faturamento é um indicador essencial, pois representa a capacidade da loja de gerar valor financeiro.

---

### 🛍 Produtos e Categorias

A Loja X apresentou forte concentração de vendas em poucos produtos, além de baixo desempenho em determinadas categorias.

Essa dependência pode indicar menor estabilidade comercial e maior risco estratégico.

---

### ⭐ Avaliação dos Clientes

A média de avaliações da Loja X foi inferior às demais unidades, sugerindo menor satisfação do cliente.

A satisfação impacta diretamente na fidelização e competitividade da loja.

---

### 🚚 Frete Médio

O custo médio de frete da Loja X foi um dos mais elevados.

Como o frete é pago pelo cliente, valores mais altos podem reduzir a atratividade da loja e influenciar negativamente a decisão de compra.

## 3️⃣ Conclusão e Recomendação

Após análise conjunta dos indicadores financeiros, operacionais e de satisfação do cliente, conclui-se que:

> ⚠️ **Recomenda-se a venda da Loja X.**

### Justificativas:

- 📉 Menor faturamento total  
- ⭐ Menor média de avaliação  
- 🚚 Maior custo médio de frete  
- 📦 Forte concentração de vendas em poucos produtos  

As demais lojas apresentam maior estabilidade financeira, melhor percepção dos clientes e maior equilíbrio comercial.

Dessa forma, a venda da Loja X representa a decisão mais estratégica para fortalecimento da rede e realocação de investimentos.
